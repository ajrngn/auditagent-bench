# 03 â€” Labeling sync

Push unlabeled candidates to the review Google Sheet; pull human-verified rows back to `processed/labels/`.

**The sheet is the labeling UI.** Do not build a custom labeling app â€” the reviewer works on a browser-only machine.

Sync is idempotent in both directions and matches on `item_id`. Re-running must not duplicate rows or clobber a reviewer's in-progress edits. The Apps Script half lives in `apps_script/Code.gs`.

Read the `labeling-controls` skill before changing anything here.

```python
# --- Config. Every tunable value in this notebook lives in this cell. ---
BUCKET = ''            # TODO: GCS bucket name
PROCESSED_PREFIX = 'processed'
SHEET_ID = ''          # TODO: review sheet ID
WORKSHEET = 'review'

# Google Sheets caps a single cell at 50,000 characters.
MAX_CELL_CHARS = 45000
```

```python
import gspread
from google.auth import default
from src import gcs

creds, _ = default()
sheet = gspread.authorize(creds).open_by_key(SHEET_ID).worksheet(WORKSHEET)

COLUMNS = ['item_id', 'accession_number', 'company_name', 'fiscal_year',
           'item_9a_text', 'severity', 'coso_component', 'is_deficient',
           'reviewed', 'reviewer_notes']

if not sheet.row_values(1):
    sheet.update('A1', [COLUMNS])
    print('wrote header row')
```

```python
# Push: add only candidates the sheet does not already have.
existing = {r['item_id'] for r in sheet.get_all_records() if r.get('item_id')}
records = list(gcs.read_jsonl(BUCKET, f'{PROCESSED_PREFIX}/item9a.jsonl'))

new_rows = []
for i, r in enumerate(records):
    item_id = f'C-{i:04d}'
    if item_id in existing:
        continue
    new_rows.append([item_id, r['accession_number'], r['company_name'],
                     r['fiscal_year'], r['item_9a_text'][:MAX_CELL_CHARS],
                     '', '', '', 'FALSE', ''])

if new_rows:
    sheet.append_rows(new_rows, value_input_option='RAW')
print(f'pushed={len(new_rows)} already_present={len(existing)}')
```

```python
# Pull: only rows a human ticked as reviewed.
#
# label_source is stamped by the Apps Script path the reviewer triggers, not here.
# Never write 'llm_draft_human_verified' programmatically â€” it asserts a human
# review that did not happen.
rows = sheet.get_all_records()
reviewed = [r for r in rows if str(r.get('reviewed')).upper() == 'TRUE']

if rows:
    print(f'{len(reviewed)} of {len(rows)} rows reviewed ({len(reviewed) / len(rows):.0%})')
else:
    print('sheet is empty')

unstamped = [r['item_id'] for r in reviewed if not r.get('label_source')]
if unstamped:
    print(f'WARNING: {len(unstamped)} reviewed rows have no label_source â€” '
          f'the Apps Script writeback did not run for them. Not exporting.')
elif reviewed:
    gcs.write_jsonl(BUCKET, f'{PROCESSED_PREFIX}/labels/reviewed.jsonl', reviewed)
    print(f'wrote {len(reviewed)} reviewed rows')
```
